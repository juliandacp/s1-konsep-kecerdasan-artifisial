## 7. Representasi Graph di Python

Cara paling umum dan gampang di Python: pakai **dictionary of list** (adjacency list).

Setiap **key** = satu node, **value**-nya = list dari node-node tetangganya.

````python
# Graph tak berarah (undirected) sederhana
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

for node, tetangga in graph.items():
    print(f"{node} -> {tetangga}")
````

Kalau digambarkan, graph di atas kira-kira bentuknya begini:

```
      A
     / \
    B   C
   /|    \
  D E     F
     \   /
      \ /
       (E-F terhubung)
```

## 8. BFS (Breadth-First Search)

Jelajahi graph **level per level**, dari node awal ke tetangga-tetangganya dulu, baru ke tetangga dari tetangga, dst. Cocok buat cari **jalur terpendek** (dalam jumlah edge) di graph tanpa bobot.

Queue (FIFO — First In First Out). Di Python kita pakai `collections.deque` biar operasi tambah/ambil di depan-belakang efisien (kalau pakai `list` biasa, `pop(0)` itu lambat / O(n)).

````python
from collections import deque

def bfs(graph, start):
    visited = set()        # buat nyimpen node yang udah dikunjungi
    queue = deque([start])  # antrian, mulai dari node awal
    urutan_kunjungan = []

    visited.add(start)

    while queue:
        node = queue.popleft()   # ambil node paling depan (FIFO)
        urutan_kunjungan.append(node)

        for tetangga in graph[node]:
            if tetangga not in visited:
                visited.add(tetangga)   # tandai begitu dimasukkan ke queue
                queue.append(tetangga)

    return urutan_kunjungan

hasil_bfs = bfs(graph, 'A')
print("Urutan kunjungan BFS dari A:", hasil_bfs)
````

**Cara kerja singkat, langkah demi langkah dari node A:**
1. Mulai dari `A`, masukkan ke queue → `queue = [A]`
2. Ambil `A`, kunjungi tetangganya `B, C` → masukkan ke queue → `queue = [B, C]`
3. Ambil `B`, kunjungi tetangganya yang belum dikunjungi `D, E` → `queue = [C, D, E]`
4. Ambil `C`, kunjungi tetangganya yang belum dikunjungi `F` → `queue = [D, E, F]`
5. Ambil `D`, gak ada tetangga baru
6. Ambil `E`, gak ada tetangga baru (F udah dikunjungi)
7. Ambil `F`, selesai

Hasil: `A, B, C, D, E, F`

## 9. DFS (Depth-First Search)

Jelajahi graph **sedalam mungkin** dulu di satu cabang, baru balik (backtrack) kalau udah mentok. Cocok buat kasus kayak cek konektivitas, topological sort, atau eksplorasi state space di AI (misal solving puzzle).

Ada 2 cara umum implementasi: **rekursif** (paling natural) dan **iteratif pakai stack (LIFO)**.

````python
# DFS versi rekursif
def dfs_rekursif(graph, node, visited=None, urutan_kunjungan=None):
    if visited is None:
        visited = set()
        urutan_kunjungan = []

    visited.add(node)
    urutan_kunjungan.append(node)

    for tetangga in graph[node]:
        if tetangga not in visited:
            dfs_rekursif(graph, tetangga, visited, urutan_kunjungan)

    return urutan_kunjungan

hasil_dfs_rekursif = dfs_rekursif(graph, 'A')
print("Urutan kunjungan DFS (rekursif) dari A:", hasil_dfs_rekursif)
````

````python
# DFS versi iteratif, pakai stack manual (list biasa, karena append/pop di ujung itu O(1))
def dfs_iteratif(graph, start):
    visited = set()
    stack = [start]     # stack (LIFO)
    urutan_kunjungan = []

    while stack:
        node = stack.pop()   # ambil elemen paling atas/terakhir (LIFO)
        if node not in visited:
            visited.add(node)
            urutan_kunjungan.append(node)
            # masukkan tetangga dalam urutan terbalik biar urutan kunjungan
            # konsisten sama versi rekursif
            for tetangga in reversed(graph[node]):
                if tetangga not in visited:
                    stack.append(tetangga)

    return urutan_kunjungan

hasil_dfs_iteratif = dfs_iteratif(graph, 'A')
print("Urutan kunjungan DFS (iteratif) dari A:", hasil_dfs_iteratif)
````

## 10. Perbandingan BFS vs DFS

| | BFS | DFS |
|---|---|---|
| Struktur data | Queue (FIFO) | Stack (LIFO) / rekursi |
| Cara jelajah | Melebar per level | Menyelam sedalam mungkin dulu |
| Cocok untuk | Jalur terpendek (unweighted graph) | Eksplorasi penuh, cek konektivitas, backtracking |
| Kompleksitas | O(V + E) | O(V + E) |

*(V = jumlah node/vertex, E = jumlah edge)*

## 11. Sorting (Pengurutan) di Python

Sorting itu salah satu algoritma paling dasar dan sering dipakai. Di Python ada cara instan pakai fungsi bawaan, tapi kita juga perlu paham cara kerja di baliknya (karena di soal ujian/tugas biasanya diminta implementasi manual, bukan tinggal panggil fungsi).

### 11.1 Cara Instan: `sorted()` dan `.sort()`

### 11.2 Bubble Sort

Bandingkan dua elemen bersebelahan, kalau urutannya salah, tukar posisinya. Ulangi terus sampai gak ada lagi yang perlu ditukar. Namanya "bubble" karena elemen terbesar "menggelembung" pelan-pelan ke ujung tiap kali satu putaran selesai.

````python
def bubble_sort(arr):
    arr = arr.copy()  # biar list aslinya gak ikut berubah
    n = len(arr)

    for i in range(n):
        tertukar = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]  # tukar, gak perlu variabel temp kayak di C
                tertukar = True
        if not tertukar:   # kalau satu putaran gak ada pertukaran, berarti udah terurut, bisa berhenti lebih awal
            break

    return arr

data = [5, 2, 9, 1, 5, 6]
print("Sebelum:", data)
print("Sesudah bubble sort:", bubble_sort(data))
````

### 11.3 Selection Sort

Di setiap putaran, cari elemen **terkecil** dari bagian yang belum terurut, lalu tukar ke posisi paling depan dari bagian yang belum terurut itu. Diulang sampai semua elemen kebagian posisi yang benar.

````python
def selection_sort(arr):
    arr = arr.copy()
    n = len(arr)

    for i in range(n):
        idx_terkecil = i
        for j in range(i + 1, n):
            if arr[j] < arr[idx_terkecil]:
                idx_terkecil = j
        arr[i], arr[idx_terkecil] = arr[idx_terkecil], arr[i]  # tukar

    return arr

data = [5, 2, 9, 1, 5, 6]
print("Sebelum:", data)
print("Sesudah selection sort:", selection_sort(data))
````

### 11.4 Perbandingan Singkat

| | Bubble Sort | Selection Sort |
|---|---|---|
| Ide dasar | Tukar elemen bersebelahan kalau salah urutan | Cari elemen terkecil, taruh di depan |
| Kompleksitas | O(n²) | O(n²) |
| Jumlah pertukaran (swap) | Bisa banyak | Maksimal n kali (lebih hemat swap) |
| Best case | O(n) kalau udah terurut (ada optimasi `tertukar`) | Tetap O(n²), gak ada shortcut |

````python
# Ruang buat kalian coba-coba latihan di sini
graph_latihan = {
    1: [2, 3],
    2: [1, 4],
    3: [1, 4],
    4: [2, 3, 5],
    5: [4]
}

# tulis kode kalian di bawah ini
````

````python
# BFS

from collections import deque

def bfs(graph, start):
    visited = set()          
    queue = deque([start])
    urutan_kunjungan = []

    visited.add(start)

    while queue:
        node = queue.popleft()  
        urutan_kunjungan.append(node)

        for tetangga in graph[node]:
            if tetangga not in visited:
                visited.add(tetangga)  
                queue.append(tetangga)

    return urutan_kunjungan

hasil_bfs = bfs(graph_latihan, 1)
print("Urutan kunjungan BFS dari 1:", hasil_bfs)
````

````python
# DFS

def dfs_rekursif(graph, node, visited=None, urutan_kunjungan=None):
    if visited is None:
        visited = set()
        urutan_kunjungan = []

    visited.add(node)
    urutan_kunjungan.append(node)

    for tetangga in graph[node]:
        if tetangga not in visited:
            dfs_rekursif(graph, tetangga, visited, urutan_kunjungan)

    return urutan_kunjungan

hasil_dfs_rekursif = dfs_rekursif(graph_latihan, 1)
print("Urutan kunjungan DFS (rekursif) dari 1:", hasil_dfs_rekursif)
````

````python
# BFS Jalur Terpendek

from collections import deque

def bfs_jalur_terpendek(graph, start, finish):
    visited = set()
    queue = deque([start])
    parent = {}

    visited.add(start)
    parent[start] = None

    while queue:
        node = queue.popleft()

        if node == finish:
            break

        for tetangga in graph[node]:
            if tetangga not in visited:
                visited.add(tetangga)
                queue.append(tetangga)
                parent[tetangga] = node

    if finish not in parent:
        return []

    jalur = []
    node = finish

    while node is not None:
        jalur.append(node)
        node = parent[node]

    jalur.reverse()

    return jalur


hasil = bfs_jalur_terpendek(graph_latihan, 1, 5)
print("Jalur terpendek dari 1 ke 5:", hasil)
````

````python
# Insertion Sort

def insertionSort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

def printArray(arr):
    for i in range(len(arr)):
        print(arr[i], end=" ")
    print()

if __name__ == "__main__":
    arr = [12, 11, 13, 5, 6]
    insertionSort(arr)
    printArray(arr)
````
